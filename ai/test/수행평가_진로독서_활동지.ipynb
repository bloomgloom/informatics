````python
# ===== 글자 수 확인 도구 (수정하지 마세요) =====
import json, os, glob, datetime

최소_글자수 = 200     # 기본 최소 글자 수 (공백 제외)
최대_글자수 = 500     # 기본 최대 글자 수 (공백 제외)
NOTEBOOK_FILE = None  # 파일을 자동으로 못 찾을 때만 파일명을 직접 입력 (예: 'notebook.ipynb')

_PLACEHOLDER = "이곳을 더블클릭하고 작성하세요"

def _노트북_경로():
    if NOTEBOOK_FILE:
        return NOTEBOOK_FILE
    g = globals()
    if "__vsc_ipynb_file__" in g:              # VS Code
        return g["__vsc_ipynb_file__"]
    s = os.environ.get("JPY_SESSION_NAME")     # Jupyter Notebook / Lab
    if s and os.path.exists(s):
        return s
    files = glob.glob("*.ipynb")               # 폴더에 노트북이 하나뿐일 때
    if len(files) == 1:
        return files[0]
    return None

def 글자수(qid, 항목="", 최소=None, 최대=None):
    최소 = 최소_글자수 if 최소 is None else 최소
    최대 = 최대_글자수 if 최대 is None else 최대
    path = _노트북_경로()
    if not path or not os.path.exists(path):
        print("❌ 노트북 파일을 찾지 못했습니다. 준비 셀의 NOTEBOOK_FILE에 파일명을 입력하고 다시 실행하세요.")
        return
    with open(path, encoding="utf-8") as f:
        nb = json.load(f)
    cells = nb["cells"]
    표식 = '글자수("' + qid + '"'
    for i, c in enumerate(cells):
        src = "".join(c["source"]) if isinstance(c["source"], list) else c["source"]
        if c["cell_type"] == "code" and 표식 in src:
            if i == 0 or cells[i-1]["cell_type"] != "markdown":
                print("❌ 바로 위 셀이 작성 셀(마크다운)이 아닙니다. 셀을 지우거나 옮기지 마세요.")
                return
            s = cells[i-1]["source"]
            text = "".join(s) if isinstance(s, list) else s
            break
    else:
        print("❌ 이 셀을 파일에서 찾지 못했습니다. 먼저 저장(Ctrl+S)했는지 확인하세요.")
        return

    저장시각 = datetime.datetime.fromtimestamp(os.path.getmtime(path)).strftime("%H:%M:%S")
    제목 = f"[{항목}] " if 항목 else ""
    if _PLACEHOLDER in text:
        print(f"{제목}⚠️ 안내 문구가 아직 남아 있습니다. 문구를 삭제하고 작성하세요.")
        return
    n = len("".join(text.split()))  # 공백·줄바꿈 제외
    칸 = 20
    채움 = min(칸, round(n / 최대 * 칸)) if 최대 else 0
    막대 = "█" * 채움 + "░" * (칸 - 채움)
    if n < 최소:
        판정 = f"❌ {최소 - n}자 부족"
    elif 최대 and n > 최대:
        판정 = f"❌ {n - 최대}자 초과"
    else:
        판정 = "✅ 기준 충족"
    범위 = f"{최소}~{최대}자" if 최대 else f"{최소}자 이상"
    print(f"{제목}글자 수(공백 제외): {n}자  |  기준 {범위}  {판정}")
    print(f"{막대}   (파일 저장 시각 {저장시각} 기준)")

print("✅ 준비 완료")
````
