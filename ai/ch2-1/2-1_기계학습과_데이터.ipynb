### ① 결측치 (Missing Value)
- 뜻 : 값이 비어 있음 → `NaN` / `NA`
- 실습 : `info()`에서 전체 344개인데 어떤 속성은 ______개만 non-null
- 처리 : 특정값·평균값으로 ______ / 행을 ______
  - 실습에서는 `df.dropna()`로 삭제

### ② 이상치 (Outlier)
- 뜻 : 잘못 입력됐거나 정상 범위에서 크게 ______ 값
- 실습 : 성별(sex)에 `MALE`/`FEMALE`이 아닌 `______` 이 섞임
- 찾기 : 상자그림(boxplot), `unique()`, 산점도
- 처리 : 제거 또는 대치

## 한눈에 정리

```
기계학습 = 데이터에서 패턴을 스스로 학습
   │
   ├─ 문제 유형 : 분류 · 예측 · 군집 · 인식 · 자연어처리 · 생성
   │
   └─ 데이터 준비
        ├─ 수집    : 공공 / 민간 / 직접   (편향·저작권 주의)
        ├─ 살펴보기 : info · describe · 시각화
        ├─ 전처리   : 결측치 · 이상치 · 정규화
        └─ 핵심 속성 : 산점도 · 상관계수
```

→ 이 흐름을 **펭귄 데이터**로 실습함 → 다음은 **지진 데이터**로 혼자 적용 (*선택 활동*)
