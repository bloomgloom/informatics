## 3️⃣ 모델링 (p.85~88)

핵심 요소의 관계를 **표, 트리, 그래프**로 구조화한다.

- **표**: 행과 열로 정리한다. 조건 비교에 알맞다.
- **트리**: 계층 관계를 나타낸다. 맨 위가 루트 노드이고 사이클이 없다.
- **그래프**: 노드와 간선으로 연결 관계를 나타낸다. 사이클을 허용한다.
