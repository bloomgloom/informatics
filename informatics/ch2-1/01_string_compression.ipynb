# 문자열 압축 직접 만들기

## 1. 런-렝스 압축

- 같은 문자가 이어지면 `개수 + 문자`로 줄이기
- 한 번만 나오면 개수 생략

| 원본 | 압축 |
|---|---|
| `COFFEEEEEE FREEE` | `CO2F6E FR3E` |

## 2. 다른 문자열 넣어 보기

- `HELLO WORLD`
