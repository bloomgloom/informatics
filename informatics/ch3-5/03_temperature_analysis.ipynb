## 단계 0: 준비 (라이브러리 설치)

코랩이 아니면 필요한 파일을 이 노트북과 같은 폴더에 둔다.

```python
import importlib, sys, subprocess, platform

packages = [
    ('pandas', 'pandas'),
    ('matplotlib.pyplot', 'matplotlib'),
]

for module_name, pip_name in packages:
    try:
        importlib.import_module(module_name)
    except ModuleNotFoundError:
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pip_name])
        importlib.invalidate_caches()

# 그래프 한글 폰트 설정
import matplotlib.pyplot as plt

if 'google.colab' in sys.modules:
    subprocess.run(['apt-get', '-qq', '-y', 'install', 'fonts-nanum'])
    import matplotlib.font_manager as fm
    fm.fontManager.addfont('/usr/share/fonts/truetype/nanum/NanumGothic.ttf')
    plt.rcParams['font.family'] = 'NanumGothic'
elif platform.system() == 'Windows':
    plt.rcParams['font.family'] = 'Malgun Gothic'
elif platform.system() == 'Darwin':
    plt.rcParams['font.family'] = 'AppleGothic'
else:
    plt.rcParams['font.family'] = 'NanumGothic'

plt.rcParams['axes.unicode_minus'] = False

# 코랩이 아니면 files.upload()가 파일을 노트북과 같은 폴더에서 찾도록 안내만 한다.
if 'google.colab' not in sys.modules:
    import types

    def upload():
        print('코랩이 아니므로 파일을 이 노트북과 같은 폴더에 두세요.')
        return {}

    sys.modules.setdefault('google', types.ModuleType('google'))
    sys.modules['google.colab'] = types.ModuleType('google.colab')
    sys.modules['google.colab'].files = types.SimpleNamespace(upload=upload)

print('준비 완료')
```
