# tkinter로 온도 조절기 만들기

## 1. 창 띄우기

- `from tkinter import *`로 불러오기
- `Tk()`로 창 만들기
- `title()`로 제목, `geometry()`로 크기 정하기
- `mainloop()`로 창 띄우기

## 2. 요소 붙이기

창에 붙이는 요소

| 요소 | 하는 일 | 예 |
|---|---|---|
| `Label` | 글자 | `Label(window, text="안녕")` |
| `Button` | 버튼 | `Button(window, text="누르기")` |
| `Entry` | 한 줄 입력 칸 | `Entry(window)` |
| `Scale` | 끌어서 고르는 막대 | `Scale(window, from_=0, to=10)` |
| `Checkbutton` | 체크 상자 | `Checkbutton(window, text="자동")` |

요소를 꾸미는 값

| 값 | 뜻 | 예 |
|---|---|---|
| `text` | 보일 글자 | `text="▲"` |
| `font` | 글꼴과 크기 | `font=("Arial", 24)` |
| `fg`, `bg` | 글자색, 배경색 | `fg="white", bg="blue"` |
| `width` | 너비(글자 수) | `width=10` |

- 요소를 만든 뒤 `pack()`으로 창에 붙이기. `pack()`을 하지 않으면 보이지 않음

## 3. 기능 연결하기

- 2-1번 코드에 이어서 쓰기
- 설정 온도를 변수 `temp`에 담고 24에서 시작하기
- `up()` 함수: `temp`를 1 올리고 글자 바꾸기
  - 함수 밖 변수를 바꾸려면 `global temp`
  - `config(text=...)`로 붙어 있는 요소의 글자 바꾸기
- 버튼에 `command=up`을 넣어 함수 연결하기

## 4. 버튼 늘리고 기능 고치기

- 3번 코드에 이어서 쓰기
- `▼` 버튼과 `down()` 함수 추가하기: 1도 내리기
- 두 함수 고치기: 18℃보다 내려가지 않고, 30℃보다 올라가지 않기

## 5. 현재 온도 붙이기

### 5-1. 작동 상태 보이기

- 4번 코드에 이어서 쓰기
- 현재 온도 `now = 26`
- `Label` 2개 더 붙이기: `현재 온도: 26℃`, 작동 상태
- `show()` 함수: 글자 3개를 모두 바꾸기
  - 현재 온도가 설정 온도보다 높으면 `냉방 중`, 낮으면 `난방 중`, 같으면 `유지`
- `up()`, `down()`에서 글자 바꾸는 줄 대신 `show()` 부르기
- `mainloop()` 위에서 `show()`를 한 번 불러 첫 화면 채우기

### 5-2. 온도 센서 붙이기

- 5-1번 코드에 이어서 쓰기
- 이 폴더의 `sensor.py`는 온도 센서 시뮬레이터(고치지 않음)

| 코드 | 하는 일 |
|---|---|
| `import sensor` | 센서 불러오기 |
| `sensor.read()` | 현재 온도 읽기 |
| `sensor.aircon(state)` | 에어컨에 `냉방 중`, `난방 중`, `유지` 알리기 |

- 냉방 중이면 현재 온도가 조금씩 내려가고, 난방 중이면 올라가고, 유지면 그대로
- `show()` 맨 끝에서 `sensor.aircon(state)` 부르기
- `measure()` 함수: `now`에 `sensor.read()` 값을 넣고 `show()` 부르기
- `window.after(1000, measure)`: 1초(1000밀리초) 뒤에 `measure()` 다시 부르기
- `now = 26` 대신 `mainloop()` 위에서 `measure()`를 한 번 부르기
