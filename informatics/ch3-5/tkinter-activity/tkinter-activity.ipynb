# tkinter로 온도 조절기 만들기

## 1. 창 띄우기

- `from tkinter import *`로 불러오기
- `Tk()`로 창 만들기
- `title()`로 제목, `geometry()`로 크기 정하기
- `mainloop()`로 창 띄우기

## 2. 글자와 버튼 넣기

- `Label`로 글자 넣기
- `Button`으로 버튼 넣기
- `pack()`으로 창에 붙이기
- 버튼을 누르면 `print`로 글자 출력하기
- `command=함수 이름`으로 버튼과 함수 연결하기

## 3. 누른 횟수 세기

- 버튼을 누를 때마다 횟수 1 늘리기
- 창의 글자를 `0번 눌렀다`, `1번 눌렀다`, …로 바꾸기
- `config(text=...)`로 글자 바꾸기

## 4. 온도 조절기 만들기

- 설정 온도 24℃에서 시작하기
- ▲ 버튼: 1도 올리기
- ▼ 버튼: 1도 내리기
- 창에 `설정 온도: 24℃`처럼 보이기

## 5. 온도 범위와 작동 상태

- 18℃보다 내려가지 않고, 30℃보다 올라가지 않기
- 현재 실내 온도는 26℃
- 설정 온도가 현재 온도보다 낮으면 `냉방 중`, 높으면 `난방 중`, 같으면 `유지`를 창에 보이기
