## 단계 0: 준비 (라이브러리 설치)

코랩이 아니면 필요한 파일을 이 노트북과 같은 폴더에 둔다.

```python
import importlib, sys, subprocess

packages = [
    ('PIL', 'pillow'),
]

for module_name, pip_name in packages:
    try:
        importlib.import_module(module_name)
    except ModuleNotFoundError:
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pip_name])
        importlib.invalidate_caches()

# 코랩이 아니면 files.upload()가 파일을 노트북과 같은 폴더에서 찾도록 안내만 한다.
if 'google.colab' not in sys.modules:
    import types

    def upload():
        print('코랩이 아니므로 파일을 이 노트북과 같은 폴더에 두세요.')
        return {}

    sys.modules.setdefault('google', types.ModuleType('google'))
    sys.modules['google.colab'] = types.ModuleType('google.colab')
    sys.modules['google.colab'].files = types.SimpleNamespace(upload=upload)

print('준비 완료')
```
