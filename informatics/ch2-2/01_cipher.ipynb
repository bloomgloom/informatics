# 암호 만들고 풀기

## 1. 카이사르 암호화

- 글자를 키만큼 뒤로 밀기
- 끝을 넘어가면 처음으로 돌아가기

| 평문 | 키 | 암호문 |
|---|---|---|
| `SECRET` | 3 | `VHFUHW` |

## 2. 카이사르 복호화

아래 암호문을 키로 풀기

- 암호문: `SBWKRQ LV IXQ`
- 키: 3

## 3. 카이사르 해독

키를 모르는 암호문 풀기

- 암호문: `LEELNV LE OLHY`

## 4. 스키테일 암호화

- 막대에 감아 쓴 글자를 풀어서 읽는 암호
- 글자의 자리만 바꾸기

| 평문 | 키 | 암호문 |
|---|---|---|
| `Secret Message` | 5 | `Stse acMgreees` |

## 5. 스키테일 복호화

아래 암호문을 키로 풀기

- 암호문: `M ARHEMF OEETSOT ECL`
- 키: 4

## 6. 스키테일 해독

키를 모르는 암호문 풀기

- 암호문: `DIE AS OT NIATEL HW!`

## 7. 미니 에니그마 암호화

- 회전자 1개와 반사판으로 글자 바꾸기
- 글자를 입력할 때마다 회전자가 한 칸씩 돌아감
- `text`와 `key`를 내가 정하기

## 8. 미니 에니그마 복호화

- 내 암호문을 **같은 코드, 같은 키**로 실행하기
- 원래 문장이 나오는지 확인하기

## 9. 미니 에니그마 해독

키를 모르는 암호문 풀기

- 암호문: `PASITQ OP IBMA`
- 힌트: 문장에 `ATTACK`이 들어 있음

## 10. 비밀 숫자 만들기 (디피-헬만)

- 비밀 숫자 `a`, `b`를 내가 정하기
- `A`, `B`는 공개해도 되는 값
- 마지막 두 숫자가 같은지 확인하기

## 11. 공개키와 개인키 만들기 (RSA)

- 서로 다른 소수 `p`, `q`를 내가 정하기 (`n`이 128보다 커야 함)
- 공개키는 알려 줘도 되고, 개인키는 나만 가지기

## 12. 공개키로 암호화

- 내 문장을 숫자로 암호화하기

## 13. 개인키로 복호화

- 원래 문장이 나오는지 확인하기

## 14. 개인키 뚫어 보기

- `n`을 소인수분해해서 `p`, `q` 찾기
- `p`, `q`를 큰 소수로 바꾸면 얼마나 오래 걸리는지 확인하기
