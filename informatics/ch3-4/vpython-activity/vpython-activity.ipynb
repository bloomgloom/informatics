## 0. 준비 (라이브러리 설치)

- 아래 셀을 실행해 VPython 설치하기
- 단계마다 셀을 실행해 파일로 저장하기
- 터미널에서 `python 파일이름.py`로 실행하기
- 브라우저 탭을 닫으면 프로그램 끝

```python
import importlib.util, sys, subprocess

packages = [
    ('vpython', 'vpython'),
    ('pkg_resources', 'setuptools<81'),
]

for module_name, pip_name in packages:
    if importlib.util.find_spec(module_name) is None:
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', pip_name])

print('라이브러리 준비 완료')
```

## 3. 지구에서 공 떨어뜨리기

- 높이 10에서 떨어뜨리기
- 지구의 중력 가속도 `g = 9.8`
- 짧은 시간 `dt = 0.01`마다 속도와 높이 바꾸기
  - 속도 = 속도 - g × dt
  - 높이 = 높이 + 속도 × dt
- 바닥에 닿으면 속도를 반대 방향으로 바꾸고 0.8배로 줄이기
- 10초 동안 보여 주기

```python
%%writefile ball3.py
```
