## IP와 포트

다른 컴퓨터의 프로그램과 이야기하려면 **어느 컴퓨터의, 어느 프로그램**인지 알아야 합니다.

| 개념 | 비유 | 예 |
|---|---|---|
| **IP 주소** | 집 주소: 어느 컴퓨터인지 | `192.168.0.23` |
| **포트 번호** | 그 집의 문 번호: 컴퓨터 안의 어느 프로그램인지 | `6000` |
| `127.0.0.1` | "우리 집": 내 컴퓨터 자신 | |

한 컴퓨터에서 여러 프로그램이 네트워크를 쓸 수 있는 것은 프로그램마다 문이 다르기 때문입니다.

- Windows: `ipconfig` → `IPv4 주소`
- Mac: `ipconfig getifaddr en0`

## 소켓, 서버, 클라이언트

**소켓**은 프로그램이 네트워크로 데이터를 주고받는 출입구입니다.

- **서버**: 문을 열어 놓고 누가 오기를 **기다리는** 쪽입니다.
- **클라이언트**: 주소와 문 번호를 알고 **찾아가는** 쪽입니다.

연결된 뒤에는 양쪽 모두 보내고 받을 수 있습니다. **연결을 먼저 거는 쪽은 언제나 클라이언트입니다.** 오늘 채팅에서는 클라이언트가 먼저 말하고 번갈아 대화하기로 약속합니다.

채팅은 겉으로 두 사람의 1:1 대화처럼 보입니다. 하지만 **그 안에는 기다리는 서버와 찾아가는 클라이언트가 있습니다.** 오늘은 내가 두 역할을 모두 해 봅니다.

| 단계 | 서버 | 클라이언트 |
|---|---|---|
| 1 | 나 | 나 |
| 2 | 나 | 선생님 |
| 3 | 선생님 | 나 |
| 4 | 나 + 다른 친구 | 나 + 다른 친구 |

**알아 두기**

- 선생님 컴퓨터를 거치는 **모든 대화는 IP, 시간과 함께 기록**되며 선생님이 볼 수 있습니다. 단계 2와 3의 답은 AI가 하며, 메시지가 AI 회사로 전달됩니다.
- 부적절한 메시지는 차단됩니다. 전화번호, 이름 같은 개인정보는 보내지 마세요.

## 실행 방법

코드는 이 노트북에 쓰고, 실행은 **터미널 창 두 개**에서 합니다. 왼쪽에 서버 창, 오른쪽에 클라이언트 창을 나란히 놓으세요.

1. **터미널 창 두 개 열기** (VS Code 안의 탭이 아니라 별도의 창입니다.)
   - Windows: 탐색기에서 이 활동지가 있는 폴더를 열고, 주소창에 `cmd`를 입력합니다. 두 번 합니다.
   - Mac: 터미널 앱을 열고 `cd `(뒤에 공백)를 입력한 뒤 폴더를 끌어다 놓고 Enter를 누릅니다. `Cmd+N`으로 새 창을 열어 한 번 더 합니다.
2. **코드 칸의 첫 줄은 지우지 않기**: 칸마다 `%%writefile server.py` 또는 `%%writefile client.py`가 적혀 있습니다. 그 아래에 코드를 쓰고 칸을 실행하면 프로그램이 돌아가는 대신 **파일로 저장**됩니다.
3. **터미널에서 실행하기**: 서버 창에 `python server.py`, 클라이언트 창에 `python client.py`를 입력합니다. (Mac은 `python3`)
4. **멈추기**: `/끝`을 입력하거나 `Ctrl+C`를 누릅니다. 멈추지 않으면 창을 닫고 다시 엽니다.
5. **코드를 고치면**: 칸을 다시 실행해 저장한 뒤 터미널에서 다시 실행합니다. 위쪽 화살표 키를 누르면 앞의 명령이 다시 나옵니다.
