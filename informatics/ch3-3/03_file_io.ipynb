## 단계 0: 준비 (라이브러리 설치)

코랩이 아니면 필요한 파일을 이 노트북과 같은 폴더에 둔다.

```python
import importlib, sys, subprocess, platform

# 코랩이 아니면 files.upload()가 파일을 노트북과 같은 폴더에서 찾도록 안내만 한다.
if 'google.colab' not in sys.modules:
    import types

    def upload():
        print('코랩이 아니므로 파일을 이 노트북과 같은 폴더에 두세요.')
        return {}

    sys.modules.setdefault('google', types.ModuleType('google'))
    sys.modules['google.colab'] = types.ModuleType('google.colab')
    sys.modules['google.colab'].files = types.SimpleNamespace(upload=upload)

# 윈도우에서도 한글 파일을 UTF-8로 읽고 쓰고, csv에 빈 줄이 생기지 않게 한다.
if platform.system() == 'Windows' and not getattr(open, 'utf8_default', False):
    import builtins
    system_open = builtins.open

    def open(file, mode='r', buffering=-1, encoding=None, errors=None, newline=None, **kwargs):
        if 'b' not in mode:
            encoding = encoding or 'utf-8'
            if newline is None and 'r' not in mode and str(file).endswith('.csv'):
                newline = ''
        return system_open(file, mode, buffering, encoding, errors, newline, **kwargs)

    open.utf8_default = True

print('준비 완료')
```
