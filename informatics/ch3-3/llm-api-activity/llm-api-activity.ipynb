## 1. 채팅에서 코드로

지금까지는 채팅 화면에 **사람이 직접** 질문을 입력하고 답을 읽었습니다. 이 일을 **프로그램이 대신**하게 할 수 있습니다. 그러면 입력을 프로그램이 정해 보내고, 받은 답을 화면에 보여 주거나 다음 처리에 활용할 수 있습니다. 예를 들어 단어 퀴즈를 만들어 내고 채점까지 하는 프로그램을 만들 수 있습니다.

채팅에서 하던 일은 코드에서 다음과 같이 바뀝니다.

| 채팅 화면에서 | 코드에서 |
|---|---|
| 로그인 | API 키 |
| 모델 고르기 | `model` |
| 질문 입력 | `messages`의 `user` 내용 |
| 보내기 버튼 | `client.messages.create(...)` |
| 화면에 나온 답 | `response.content[0].text` |
| 이어지는 대화 기록 | `messages` 목록에 직접 쌓기 |
| 맞춤 지시 설정 | `system` |

### API와 API 키

- **API**: 프로그램이 다른 프로그램의 기능을 쓰는 창구입니다. 사람이 쓰는 창구가 채팅 화면이라면, 프로그램이 쓰는 창구가 Claude API입니다.
- **API 키**: 프로그램에는 로그인 화면이 없으므로, 요청마다 키를 함께 보내 누가 보낸 요청인지 알립니다. 사용 요금도 이 키의 주인에게 청구됩니다. 그래서 키는 **비밀번호와 같습니다.** 코드에 그대로 적거나 다른 사람에게 알려 주면 안 됩니다.

## 2. Claude API 사용 방법

보통 다음 순서로 사용합니다.

1. [Claude Console](https://platform.claude.com/)에 가입합니다.
2. 크레딧을 충전합니다. API는 사용한 만큼 요금이 나갑니다.
3. **API Keys** 메뉴에서 API 키를 발급받습니다.
4. 파이썬에 `anthropic` 라이브러리를 설치하고, 환경 변수 `ANTHROPIC_API_KEY`에 키를 저장한 뒤 코드를 작성합니다.

더 알아보기: [시작하기](https://platform.claude.com/docs/ko/get-started) · [Messages API](https://platform.claude.com/docs/ko/api/messages) · [모델 안내](https://platform.claude.com/docs/ko/models/overview) · [프롬프트 작성 안내](https://platform.claude.com/docs/ko/build-with-claude/prompt-engineering/overview)
