## 1️⃣ 순차 탐색과 이진 탐색 (p.103~104)

- **순차 탐색**: 처음부터 차례대로 비교한다. 정렬하지 않아도 된다.
- **이진 탐색**: 가운데 값과 비교해 범위를 절반씩 줄인다. 먼저 **정렬**해야 한다. M = (L + R) // 2
