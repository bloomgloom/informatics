## 1️⃣ 알고리즘 설계와 표현 (p.93~94)

**알고리즘**은 문제를 해결하는 절차로, 입력-처리-출력으로 이루어진다.

- 제어 구조: **순차, 선택, 반복**
- 표현 방법: **자연어, 의사 코드, 순서도**

## 2️⃣ 효율성 분석 (p.95)

효율적인 알고리즘은 수행 시간이 짧고 기억 장소를 적게 쓴다. 1부터 100까지의 합을 순서대로 더하면 덧셈이 100번쯤 필요하고, 공식(101 × 100 ÷ 2)을 쓰면 연산 세 번이면 된다.

### 🙌 활동 6

```
시각, 기온 입력
만약 오전 9 <= 시각 < 18
  LED OFF
  (빈칸)
아니면
  그늘막 접기
  LED ON
```
