## 2️⃣ 버블 정렬과 선택 정렬 (p.98~99)

- **버블 정렬**: 이웃한 두 값을 비교해 자리를 바꾼다. 단계마다 가장 큰 값이 뒤에 고정된다.
- **선택 정렬**: 남은 값 중 가장 작은 값을 찾아 기준 위치의 값과 바꾼다.
